```python
# getting the link from api

import requests
import pandas as pd
import json

url = "https://datamall2.mytransport.sg/ltaodataservice/EVCBatch"

payload = {}
headers = {
  'Accept': 'application/json',
  'accountkey': '' # input your LTA datamall API account key
}

response = requests.request("GET", url, headers=headers, data=payload)

print(response.text)

print("✅ Libraries imported!")
print(f"   requests version : {requests.__version__}")
print(f"   pandas version   : {pd.__version__}")
```

```python
# Convert JSON response to Python dictionary

data = response.json()

print(f"Type of data       : {type(data)}")
print(f"Top-level keys     : {list(data.keys())}")
```

```python
# Inspect the nested structure
# This API uses 'value' as the key containing records

print(f"Number of records  : {len(data.get('value', []))}") 
print()

# Preview first record
# json.dumps() method can convert a Python object into a JSON string.
if data.get('value'):
    print(json.dumps(data['value'][0]["Link"]))
```

```python
# getting rid of " in the url

new_url = json.dumps(data['value'][0]['Link']).strip('"')
new_url
```

```python
response1 = requests.request("GET", new_url, headers=headers, data=payload)

print(response1.text[:300])
```

```python
# Inspect the nested structure
# This API uses 'evLocationsData' as the key containing records

new_data = response1.json()

print(f"Type of data       : {type(new_data)}")
print(f"Top-level keys     : {list(new_data.keys())}")
```

```python
# checking number of records
print(f"Number of records  : {len(new_data.get('evLocationsData', []))}") 
print()

# Preview first record
# json.dumps() method can convert a Python object into a JSON string.
if new_data.get('evLocationsData'):
    print("First record (formatted):")
    print(json.dumps(new_data['evLocationsData'][0], indent=2))
    print("Second record (formatted):")
    print(json.dumps(new_data['evLocationsData'][1], indent=2))
```

```python
# Use pd.json_normalize to flatten 'evLocationsData' → 'attributes' into a DataFrame

df_location = pd.json_normalize(new_data, 'evLocationsData')

print(f"Shape : {df_location.shape}")
df_location.head(5)
```

```python
# Use pd.json_normalize to flatten 'chargingPoints' → 'attributes' into a DataFrame

df_chargingpoints = pd.json_normalize(new_data['evLocationsData'], 'chargingPoints')

print(f"Shape : {df_chargingpoints.shape}")
df_chargingpoints.head(5)
```

```python
# Use pd.json_normalize to flatten 'plugTypes' → 'attributes' into a DataFrame

dfs_plugtype = []

for i in range(len(new_data['evLocationsData'])):
    for j in range(len(new_data['evLocationsData'][i]['chargingPoints'])):
        df1 = pd.json_normalize(
            new_data['evLocationsData'][i]['chargingPoints'][j],
            'plugTypes'
        )
        dfs_plugtype.append(df1)

df_plugtype = pd.concat(dfs_plugtype, ignore_index=True)

print(f"Shape : {df_plugtype.shape}")
df_plugtype.head(5)
```

```python
# Use pd.json_normalize to flatten 'evIds' → 'attributes' into a DataFrame

dfs_evids = []

for i in range(len(new_data['evLocationsData'])):
    for j in range(len(new_data['evLocationsData'][i]['chargingPoints'])):
        for k in range(len(new_data['evLocationsData'][i]['chargingPoints'][j]['plugTypes'])):
            df2 = pd.json_normalize(
                new_data['evLocationsData'][i]['chargingPoints'][j]['plugTypes'][k],
                'evIds'
                )
            dfs_evids.append(df2)

df_evids = pd.concat(dfs_evids, ignore_index=True)

print(f"Shape : {df_evids.shape}")
#df_evids.head(5)
```

```python
# use for loop to flatten out all the nested structure and combine all into 1 big table

records = []

for location in new_data['evLocationsData']:
    
    for charging_point in location['chargingPoints']:
        
        for plug_type in charging_point['plugTypes']:
            
            for ev_id in plug_type['evIds']:
                
                record = {
                    'address': location['address'],
                    'location_name': location['name'],
                    'longtitude': location['longtitude'],
                    'latitude': location['latitude'],
                    'postal_code': location['postalCode'],
                    
                    'chargingpoint_status': charging_point['status'],
                    'operating_hours': charging_point['operatingHours'],
                    'operator': charging_point['operator'],
                    'position': charging_point['position'],
                    'chargingpoint_name': charging_point['name'],
                    
                    'plug_type': plug_type['plugType'],
                    'price': plug_type['price'],
                    'current': plug_type['current'],
                    'power_rating': plug_type['powerRating'],
                    'price_type': plug_type['priceType'],
                    
                    'evcp_id': ev_id['evCpId'],
                    'evid_status': ev_id['status']
                }
                
                records.append(record)

df_main = pd.DataFrame(records)

print(f"Shape : {df_main.shape}")
df_main.head(5)
```

```python
# converting to a more appropriate data type for each column

df_main.postal_code = df_main.postal_code.astype(int)
df_main.chargingpoint_status = df_main.chargingpoint_status.astype(int)
df_main.price = pd.to_numeric(df_main.price, errors='coerce')
df_main.evid_status = pd.to_numeric(df_main.evid_status, errors='coerce')
```

```python
# === CONFIGURE postgresql CONNECTION ===

USERNAME = '' 
PASSWORD = '' 
HOST     = '' 
PORT     = 
DB_NAME  = 'ev'

CONNECTION_STRING = f'postgresql://{USERNAME}:{PASSWORD}@{HOST}:{PORT}/{DB_NAME}'
print(f"Connection string: postgresql://{USERNAME}:****@{HOST}:{PORT}/{DB_NAME}")
```

```python
# Create a new database (one-time operation)
engine = db.create_engine(CONNECTION_STRING)

if not database_exists(engine.url):
    create_database(engine.url)
    print(f"✅ Database '{DB_NAME}' created successfully!")
else:
    print(f"ℹ️  Database '{DB_NAME}' already exists — skipping creation.")

# Always dispose engine after use to free resources
engine.dispose()
print("Engine disposed (resources released).")
```

```python
# i realised this step is not needed to load the dataframe straight into postgresql.
# can straight execute this line: df_main.to_sql(name='ev_main', con=engine, if_exists='append', index=False) 

commands = (
    # Table 1: ev_main — stores ev charing point data
    """
    DROP TABLE IF EXISTS ev_main;
    CREATE TABLE ev_main (
        address               TEXT,
        location_name         TEXT,
        longtitude            DOUBLE PRECISION,
        latitude              DOUBLE PRECISION,
        postal_code           INTEGER,
        chargingpoint_status  SMALLINT,
        operating_hours       TEXT,
        operator              TEXT,
        position              TEXT,
        chargingpoint_name    TEXT,
        plug_type             TEXT,
        price                 NUMERIC,
        current               VARCHAR(3),
        power_rating          NUMERIC,
        price_type            VARCHAR(5),
        evcp_id               TEXT,
        evid_status           SMALLINT
    );
    """,

    
)

# Execute all commands within a single transaction
with engine.begin() as conn:
    for command in commands:
        conn.execute(text(command))

print("✅ Tables 'ev_main' created successfully!")
```
